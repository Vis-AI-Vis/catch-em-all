# Pokémon classifier

Model: [skshmjn/Pokemon-classifier-gen9-1025](https://huggingface.co/skshmjn/Pokemon-classifier-gen9-1025) (ViT, 1025 Pokémon classes)

We compare the predictions on four datasets:
- **pokemon**: images of classes the model was trained on
- **animals**: real animals, which are *not* classes of the model (out-of-distribution)
- **am2k_original**: AM-2k, 2000 real animal photos from 20 categories (out-of-distribution)
- **am2k_bg_removed**: the same AM-2k photos, cut out with the dataset's masks and placed on a white background like the Pokémon images (out-of-distribution)


Pokemon Datasets: 
Pokemon Dataset: https://www.kaggle.com/datasets/hlrhegemony/pokemon-image-dataset

OOD Datasets:
Animal Dataset https://github.com/bruinxiong/animal-matting
AM2K: https://drive.google.com/drive/folders/1SReB9Zma0TDfDhow7P5kiZNMwY9j9xMA 

Predictions are saved to `predictions/` and only computed again if those files are missing.

`load_model()` downloads the model from Hugging Face on the first run (about 330 MB) and caches it in `models/` (`MODELS_DIR` in `src/pokemon_classifier.py`). Later runs load it from there without downloading it again. To download it again, delete `models/`.

## Load model and datasets

````python
model, image_processor = load_model()
id2label = model.config.id2label
print(f"Number of classes: {model.config.num_labels}")

datasets = load_datasets(model, image_processor)
for dataset in datasets:
    print(f"{dataset.name}: {len(dataset)} images")
````

## Predictions

The first run predicts every image, which takes a while on CPU. Later runs load the saved files.

- `predictions[name]`: one row per image (path, true and predicted label, confidence, entropy)
- `confidences[name]`: full probability matrix, shape `(num_images, 1025)`. Row `i` belongs to row `i` of the DataFrame.

````python
predictions = {}
confidences = {}
for dataset in datasets:
    predictions[dataset.name], confidences[dataset.name] = load_or_predict(
        dataset, model, image_processor
    )

pokemon_predictions_df = predictions["pokemon"]
animal_predictions_df = predictions["animals"]
am_2k_original_predictions_df = predictions["am2k_original"]
am_2k_bg_removed_predictions_df = predictions["am2k_bg_removed"]
#pokemon_predictions_df.head()
````

## Examples

Images with their prediction, e.g. the animals the model is most confident about, or Pokémon it got wrong.

````python
from src.pokemon_classifier import POKEMON_DIR


def pokemon_reference_image(label):
    """First image of a Pokémon from the Pokémon dataset, or None if there is none.

    The model knows 1025 Pokémon, the dataset only has images of some of them.
    """
    folder = POKEMON_DIR / label
    if not folder.is_dir():
        return None
    paths = sorted(p for p in folder.iterdir() if p.suffix.lower() in {".png", ".jpg", ".jpeg", ".webp"})
    return Image.open(paths[0]).convert("RGB") if paths else None


def show_images(df, n=8, cols=None, show_prediction=True):
    """Show the first n images of a predictions DataFrame with their prediction.

    show_prediction: put an image of the predicted Pokémon next to each image,
                     to compare them visually
    cols: images (or image + Pokémon pairs) per row
    """
    df = df.head(n)
    per_item = 2 if show_prediction else 1
    if cols is None:
        cols = 3 if show_prediction else 4
    cols = min(cols, len(df))
    rows = int(np.ceil(len(df) / cols))
    fig, axes = plt.subplots(
        rows, cols * per_item, figsize=(2.6 * cols * per_item, 3.0 * rows), squeeze=False
    )
    for ax in axes.flat:
        ax.axis("off")

    for i, (_, row) in enumerate(df.iterrows()):
        r, c = divmod(i, cols)
        ax = axes[r, c * per_item]
        ax.imshow(Image.open(row["image_path"]).convert("RGB"))
        ax.set_title(
            f"true: {row['true_label']}\n"
            f"pred: {row['predicted_label']} ({row['confidence']:.2f})",
            fontsize=9,
        )
        if show_prediction:
            ax = axes[r, c * per_item + 1]
            reference = pokemon_reference_image(row["predicted_label"])
            if reference is not None:
                ax.imshow(reference)
            else:
                ax.text(0.5, 0.5, "no image in\nPokémon dataset", ha="center", va="center",
                        fontsize=8, color="gray", transform=ax.transAxes)
            ax.set_title(f"→ {row['predicted_label']}", fontsize=9, color="dimgray")
    fig.tight_layout()
    plt.show()
````

````python
# Animals the model is most confident about (most "Pokémon-like")
for name in ["animals", "am2k_original", "am2k_bg_removed"]:
    print(f"Dataset: {name}")
    show_images(predictions[name].sort_values("confidence", ascending=False))
````

````python
# Wrong Pokémon predictions, most confident first
wrong = pokemon_predictions_df[pokemon_predictions_df["true_label_id"] != pokemon_predictions_df["predicted_label_id"]]
print(f"{len(wrong)} wrong predictions")
show_images(wrong.sort_values("confidence", ascending=False))
````

## Top-5 predictions of a single image

````python
def top_k(dataset_name, index, k=5):
    """Top-k classes and probabilities of one image."""
    probs = confidences[dataset_name][index]
    top = np.argsort(probs)[::-1][:k]
    return pd.DataFrame({"label": [id2label[i] for i in top], "probability": probs[top]})


# Most confident image of each animal dataset and its top-5 Pokémon
for name in ["animals", "am2k_original", "am2k_bg_removed"]:
    dataset_predictions = predictions[name]
    index = dataset_predictions["confidence"].idxmax()

    print(f"Dataset: {name}")
    show_images(dataset_predictions.loc[[index]], n=1, cols=1)
    print(top_k(name, index))
    print("\n")
````

## Confident predictions on animals (confidence > 70 %)

The animal datasets have no correct Pokémon class, so every confident prediction is a confident mistake. Which animals cause them, and which Pokémon do they get mapped to?

````python
THRESHOLD = 0.7
ANIMAL_DATASETS = ["animals", "am2k_original", "am2k_bg_removed", "am2k_bg_removed_cartoonized"]

confident = {
    name: predictions[name][predictions[name]["confidence"] > THRESHOLD]
    for name in ANIMAL_DATASETS
}

pd.DataFrame(
    {
        "images": [len(predictions[name]) for name in ANIMAL_DATASETS],
        f"confidence > {THRESHOLD:.0%}": [len(confident[name]) for name in ANIMAL_DATASETS],
        "share": [len(confident[name]) / len(predictions[name]) for name in ANIMAL_DATASETS],
        "mean entropy_norm": [confident[name]["entropy_norm"].mean() for name in ANIMAL_DATASETS],
    },
    index=ANIMAL_DATASETS,
).style.format({"share": "{:.1%}", "mean entropy_norm": "{:.3f}"})
````

````python
# Most frequent (animal -> Pokémon) pairs among the confident predictions
for name in ANIMAL_DATASETS:
    pairs = (
        confident[name]
        .groupby(["true_label", "predicted_label"])
        .agg(count=("confidence", "size"), mean_confidence=("confidence", "mean"))
        .sort_values("count", ascending=False)
    )
    print(f"{name}: {len(pairs)} different pairs")
    display(pairs.head(10))
````

````python
# One example per (animal -> Pokémon) pair, most frequent pairs first
for name in ANIMAL_DATASETS:
    df = confident[name]
    pair_counts = df.groupby(["true_label", "predicted_label"])["confidence"].transform("size")
    examples = (
        df.assign(pair_count=pair_counts)
        .sort_values(["pair_count", "confidence"], ascending=False)
        .drop_duplicates(["true_label", "predicted_label"])
    )
    print(f"Dataset: {name}")
    show_images(examples, n=8)
````

### AM-2k: does removing the background change which images are confident?

Both AM-2k datasets contain the same photos, so we can check for each photo whether it is above the threshold in the original, after background removal, or in both.

````python
def with_file(df):
    return df.assign(file=df["image_path"].map(lambda path: path.split("/")[-1]))


am2k = pd.merge(
    with_file(predictions["am2k_original"]),
    with_file(predictions["am2k_bg_removed"]),
    on=["file", "true_label"],
    suffixes=("_original", "_bg_removed"),
)
above_original = am2k["confidence_original"] > THRESHOLD
above_bg_removed = am2k["confidence_bg_removed"] > THRESHOLD

print(f"Confident in both:              {(above_original & above_bg_removed).sum()}")
print(f"Only in the original:           {(above_original & ~above_bg_removed).sum()}")
print(f"Only after background removal:  {(~above_original & above_bg_removed).sum()}")

# Photos that are confident in only one version: what did the background change?
changed = am2k[above_original != above_bg_removed]
changed[
    ["true_label", "predicted_label_original", "confidence_original",
     "predicted_label_bg_removed", "confidence_bg_removed"]
].sort_values("true_label")
````

### AM-2k (background removed): animal class vs. predicted Pokémon

Confusion matrix of the confident predictions (confidence > 70 %): true animal class on the y axis, predicted Pokémon on the x axis, each cell is the number of images with that combination. Rows and columns are sorted by their total count.

````python
def plot_confusion_matrix(df, title, all_classes=None, save_path=None):
    """Heatmap of true animal class (rows) vs. predicted Pokémon (columns), cells = counts.

    all_classes: if given, every one of these classes gets a row, even with no images.
    """
    matrix = pd.crosstab(df["true_label"], df["predicted_label"])
    if all_classes is not None:
        matrix = matrix.reindex(sorted(all_classes), fill_value=0)
    # Most frequent animals at the top (ties alphabetical), most frequent Pokémon on the left
    matrix = matrix.loc[
        matrix.sum(axis=1).sort_values(ascending=False, kind="stable").index,
        matrix.sum(axis=0).sort_values(ascending=False).index,
    ]

    fig, ax = plt.subplots(
        figsize=(1.0 * matrix.shape[1] + 3, 0.45 * matrix.shape[0] + 2)
    )
    image = ax.imshow(matrix.values, cmap="Blues", vmin=0, aspect="auto")

    # Label the non-empty cells with their count
    threshold = matrix.values.max() / 2
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            count = matrix.values[i, j]
            if count > 0:
                ax.text(
                    j, i, count, ha="center", va="center", fontsize=10,
                    color="white" if count > threshold else "black",
                )

    ax.set_xticks(range(matrix.shape[1]), matrix.columns, rotation=45, ha="right")
    ax.set_yticks(range(matrix.shape[0]), matrix.index)
    ax.set_xlabel("Predicted Pokémon")
    ax.set_ylabel("True animal class")
    ax.set_title(title)
    # Thin white grid lines between the cells
    ax.set_xticks(np.arange(matrix.shape[1] + 1) - 0.5, minor=True)
    ax.set_yticks(np.arange(matrix.shape[0] + 1) - 0.5, minor=True)
    ax.grid(which="minor", color="white", linewidth=2)
    ax.tick_params(which="minor", length=0)
    for spine in ax.spines.values():
        spine.set_visible(False)
    fig.colorbar(image, ax=ax, label="Number of images")
    fig.tight_layout()
    if save_path is not None:
        fig.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show()
    return matrix


am2k_bg_removed_matrix = plot_confusion_matrix(
    confident["am2k_bg_removed"],
    f"AM-2k (background removed), confidence > {THRESHOLD:.0%} "
    f"({len(confident['am2k_bg_removed'])} of {len(am_2k_bg_removed_predictions_df)} images)",
    all_classes=predictions["am2k_bg_removed"]["true_label"].unique(),
    save_path=PREDICTIONS_DIR / "am2k_bg_removed_confusion_matrix.png",
)
am2k_bg_removed_matrix
````

## Cartoonized AM-2k dataset

Makes the background-removed AM-2k photos look more like the Pokémon artwork (flat colours, vivid colours, dark outlines), to test whether this increases the model's confidence. The parameters are explained in `cartoonize_image` in `src/preprocessing.py`.

The images are saved to `images/AM-2k_bg-removed-cartoonized/<animal>/`. Images that already exist are skipped; set `overwrite=True` after changing the parameters.

````python
from src.preprocessing import create_cartoon_dataset
from src.pokemon_classifier import AM2K_BG_REMOVED_DIR, IMAGES_DIR

AM2K_CARTOON_DIR = IMAGES_DIR / "AM-2k_bg-removed-cartoonized"

create_cartoon_dataset(
    AM2K_BG_REMOVED_DIR,
    AM2K_CARTOON_DIR,
    #overwrite=True  # True: create all images again, e.g. after changing the parameters
    # Parameters of cartoonize_image (these are the defaults)
    max_size=384,
    smoothing=6,
    colours=6,
    cleanup=9,
    saturation=1.8,
    brightness=1.25,
    outline_thresholds=(80, 180),
    min_line_length=25,
    outline_width=2,
    silhouette_width=3,
    background_threshold=235,
)
````

````python
from src.data import PokemonDataset

AM2K_DATASETS = ["am2k_original", "am2k_bg_removed", "am2k_bg_removed_cartoonized"]

cartoon_dataset = PokemonDataset(
    AM2K_CARTOON_DIR, image_processor, label2id=None, name="am2k_bg_removed_cartoonized"
)
print(f"{cartoon_dataset.name}: {len(cartoon_dataset)} images")

# Saved predictions from an earlier run with fewer images (e.g. the [::20] test) are outdated
saved_csv = PREDICTIONS_DIR / f"{cartoon_dataset.name}_predictions.csv"
if saved_csv.exists() and len(pd.read_csv(saved_csv)) != len(cartoon_dataset):
    print("Saved predictions don't match the images, predicting again")
    saved_csv.unlink()

predictions[cartoon_dataset.name], confidences[cartoon_dataset.name] = load_or_predict(
    cartoon_dataset, model, image_processor
)

# Overview, as for the other datasets
summary = pd.concat({name: predictions[name] for name in AM2K_DATASETS}, names=["dataset"])
display(summary.groupby("dataset")[["confidence", "entropy_norm"]].describe().T)

fig = visualise_confidence_vs_entropy_interactive(
    {name: predictions[name] for name in AM2K_DATASETS},
    filterable=AM2K_DATASETS,
    save_path=PREDICTIONS_DIR / "confidence_vs_entropy_am2k_cartoonized.html",
)

# Cartoonized animals the model is most confident about
show_images(predictions["am2k_bg_removed_cartoonized"].sort_values("confidence", ascending=False))
````

````python
# Confusion matrix of the confident cartoonized predictions, like the one for am2k_bg_removed above
cartoon_predictions_df = predictions["am2k_bg_removed_cartoonized"]
confident["am2k_bg_removed_cartoonized"] = cartoon_predictions_df[
    cartoon_predictions_df["confidence"] > THRESHOLD
]

am2k_cartoon_matrix = plot_confusion_matrix(
    confident["am2k_bg_removed_cartoonized"],
    f"AM-2k (background removed, cartoonized), confidence > {THRESHOLD:.0%} "
    f"({len(confident['am2k_bg_removed_cartoonized'])} of {len(cartoon_predictions_df)} images)",
    all_classes=cartoon_predictions_df["true_label"].unique(),
    save_path=PREDICTIONS_DIR / "am2k_bg_removed_cartoonized_confusion_matrix.png",
)
am2k_cartoon_matrix
````
